Handling missing values

### Evaluating Missing Values with '-' as Missing Data

As suggested, we will now consider '-' as a missing value and replace it with `NaN` across the DataFrame before re-checking for missing values.

```python
# Replace all '-' values with NaN in the DataFrame
df.replace('-', pd.NA, inplace=True)

# Re-calculate missing values after replacing '-' with NaN
missing_values_rechecked = df.isnull().sum()
missing_values_rechecked = missing_values_rechecked[missing_values_rechecked > 0].sort_values(ascending=False)

print("Columns with missing values (including '-' as NaN) and their counts:")
display(missing_values_rechecked)

# Display the DataFrame info again to see updated counts
display(df.info())
```

### Droping irrelevant columns based on Threshold

we will now drop columns that have more than 95% missing values, with the exception of `gnodeb_id` which is to be retained. For the remaining columns with missing values, including `gnodeb_id`, we will fill them appropriately.


`gnodeb_id` is required for 5G connection detections therefore it is retained.

```python
total_rows = df.shape[0]
missing_percentage = (df.isnull().sum() / total_rows) * 100

# Identify columns to drop (more than 95% missing) but exclude 'gnodeb_id'
columns_to_drop = missing_percentage[missing_percentage > 95].index.tolist()

if 'gnodeb_id' in columns_to_drop:
    columns_to_drop.remove('gnodeb_id')
    print("Retaining 'gnodeb_id' as per user request, even though it has >95% missing values.")

if columns_to_drop:
    df.drop(columns=columns_to_drop, inplace=True)
    print(f"Dropped columns with >95% missing values (excluding gnodeb_id): {columns_to_drop}")
else:
    print("No columns to drop based on the >95% missing value threshold (excluding gnodeb_id).")

# Now, re-evaluate missing values and fill them for remaining columns
# For object columns, fill with 'Unknown'
for col in df.select_dtypes(include=['object']).columns:
    if df[col].isnull().any():
        df[col].fillna('Unknown', inplace=True)

# For other types (e.g., numeric), we might need different strategies, but for now assuming object types cover the rest.
# If there are datetime columns with NaT, they will remain as such, which is acceptable.

print("\nMissing values after dropping columns and filling object types:")
missing_values_final = df.isnull().sum()
missing_values_final = missing_values_final[missing_values_final > 0].sort_values(ascending=False)

if missing_values_final.empty:
    print("All missing values have been handled.")
else:
    display(missing_values_final)

display(df.head())
```

### Fixing Data Types for Memory Optimization

To improve memory efficiency and ensure appropriate data representation, we will now optimize the data types of the DataFrame columns. This involves:

1.  **Downcasting Numerical Columns:** Converting integer columns to the smallest possible integer type (e.g., `int64` to `int32`, `int16`, or `int8`) if their value range permits.
2.  **Converting Object Columns to Categorical:** Transforming `object` type columns that are essentially categorical into `category` dtype. This is particularly useful for columns that were kept as is and were not explicitly label-encoded (e.g., `location_information`, `additional_information`, `enodeb_id`, `gnodeb_id`).

```python
# Downcast numerical columns to more memory-efficient types
for col in df.select_dtypes(include=['int64', 'int32']).columns:
    # Using pd.to_numeric with downcast='integer' is a safe way to do this
    # It will only downcast if no precision is lost
    df[col] = pd.to_numeric(df[col], downcast='integer')

print("Numerical columns downcasted for memory optimization.")

# Convert remaining 'object' type columns (that are not intended for further string processing)
# to 'category' dtype for memory efficiency
object_cols_to_convert_to_category = [
    'location_information',
    'additional_information',
    'enodeb_id',
    'gnodeb_id'
]

for col in object_cols_to_convert_to_category:
    if col in df.columns:
        # Ensure we only convert if the column is currently of object type
        if df[col].dtype == 'object':
            df[col] = df[col].astype('category')
            print(f"Column '{col}' converted to 'category' type.")
        else:
            print(f"Column '{col}' is not of 'object' type, skipping category conversion.")
    else:
        print(f"Column '{col}' not found, skipping category conversion.")

print("\nDataFrame info after data type optimization:")
display(df.info())
display(df.head())
```

#### 10. `alarm_duration` -> Convert to numeric minutes

```python
def convert_duration_to_minutes(duration_str):
    if pd.isna(duration_str) or duration_str == 'Unknown': # Handle NaN and 'Unknown' values
        return np.nan # Use np.nan for missing/unknown durations

    total_minutes = 0
    parts = duration_str.split(' ')
    i = 0
    while i < len(parts):
        try:
            value = int(parts[i])
            unit = parts[i+1].lower()
            if 'hour' in unit:
                total_minutes += value * 60
            elif 'minute' in unit:
                total_minutes += value
            elif 'second' in unit:
                total_minutes += value / 60
            i += 2
        except (ValueError, IndexError):
            i += 1 # Skip problematic parts
            continue
    return total_minutes

if 'alarm_duration' in df.columns:
    df['alarm_duration_minutes'] = df['alarm_duration'].apply(convert_duration_to_minutes)
    df.drop(columns=['alarm_duration'], inplace=True)
    print("'alarm_duration' converted to 'alarm_duration_minutes'.")
else:
    print("'alarm_duration' column not found, skipping conversion.")
```

### Feature Engineering

Based on the project objectives, we will now engineer sequence-based features per tower (`enodeb_id`). Each tower's alarms will be processed independently and chronologically. The `WINDOW_SIZE` parameter will control how many previous alarms are considered for certain features.

```python
# Define the window size for previous alarms
WINDOW_SIZE = 5

print(f"Feature engineering will use a window size of: {WINDOW_SIZE}")
```

First, we must ensure the DataFrame is sorted by `enodeb_id` and then chronologically by the `occurred_on` attributes (hour, day, weekday, month) for correct sequence-based feature generation. Since the original `occurred_on` datetime column was dropped, we will sort using the extracted temporal features, acknowledging this might not be perfectly precise down to the second/minute.

```python
# Sort the DataFrame by enodeb_id and then by occurred_on's temporal features
df.sort_values(by=['enodeb_id', 'occurred_on_month', 'occurred_on_day', 'occurred_on_hour', 'occurred_on_weekday'], inplace=True)
print("DataFrame sorted by 'enodeb_id' and 'occurred_on' temporal features.")
display(df.head())
```

#### 1. Previous Alarm Features (`previous_alarm_N`)

We will create features indicating the `alarm_name_encoded` of the `N` most recent previous alarms for each tower. A value of -1 will indicate no previous alarm.

```python
# Generate previous_alarm_N features
for i in range(1, WINDOW_SIZE + 1):
    df[f'previous_alarm_{i}'] = df.groupby('enodeb_id')['alarm_name_encoded'].shift(i).fillna(-1).astype(int)
print(f"Generated 'previous_alarm_1' to 'previous_alarm_{WINDOW_SIZE}' features.")
display(df.head())
```

#### 2. Previous Severity Features (`previous_severity_N`)

Similarly, we create features for the `severity_encoded` of the `N` most recent previous alarms for each tower.

```python
# Generate previous_severity_N features
for i in range(1, WINDOW_SIZE + 1):
    df[f'previous_severity_{i}'] = df.groupby('enodeb_id')['severity_encoded'].shift(i).fillna(-1).astype(int)
print(f"Generated 'previous_severity_1' to 'previous_severity_{WINDOW_SIZE}' features.")
display(df.head())
```

#### 3. Previous Alarm Source Features (`previous_alarm_source_N`)

Generating features for the `alarm_source_encoded` of the `N` most recent previous alarms.

```python
# Generate previous_alarm_source_N features
for i in range(1, WINDOW_SIZE + 1):
    df[f'previous_alarm_source_{i}'] = df.groupby('enodeb_id')['alarm_source_encoded'].shift(i).fillna(-1).astype(int)
print(f"Generated 'previous_alarm_source_1' to 'previous_alarm_source_{WINDOW_SIZE}' features.")
display(df.head())
```

#### 4. Previous Object Type Features (`previous_object_type_N`)

Generating features for the `object_type_encoded` of the `N` most recent previous alarms.

```python
# Generate previous_object_type_N features
for i in range(1, WINDOW_SIZE + 1):
    df[f'previous_object_type_{i}'] = df.groupby('enodeb_id')['object_type_encoded'].shift(i).fillna(-1).astype(int)
print(f"Generated 'previous_object_type_1' to 'previous_object_type_{WINDOW_SIZE}' features.")
display(df.head())
```

#### 9. Previous Cell Unavailable Count (`previous_cell_unavailable_count`)

This feature counts how many 'Cell Unavailable' alarms have occurred previously for a given tower.

```python
# Get the encoded value for 'Cell Unavailable'
cell_unavailable_encoded_value = label_encoders['alarm_name'].transform(['Cell Unavailable'])[0]

# Create a boolean series for 'Cell Unavailable' alarms
df['is_cell_unavailable'] = (df['alarm_name_encoded'] == cell_unavailable_encoded_value).astype(int)

# Calculate previous_cell_unavailable_count
# Use cumsum() and then shift(1) to get the count *before* the current alarm
df['previous_cell_unavailable_count'] = df.groupby('enodeb_id')['is_cell_unavailable'].cumsum().shift(1).fillna(0).astype(int)

# Drop the intermediate 'is_cell_unavailable' column
df.drop(columns=['is_cell_unavailable'], inplace=True)

print("Generated 'previous_cell_unavailable_count' feature.")
display(df.head())
```

### Target Variable Creation

We will create a binary target variable (`target`) indicating whether a 'Cell Unavailable' event occurs *after* the current alarm within the same tower's sequence.

```python
# Recreate 'is_cell_unavailable' for target creation
cell_unavailable_encoded_value = label_encoders['alarm_name'].transform(['Cell Unavailable'])[0]
df['is_cell_unavailable_for_target'] = (df['alarm_name_encoded'] == cell_unavailable_encoded_value).astype(int)

# Calculate the target: 1 if a 'Cell Unavailable' occurs AFTER the current alarm in the same tower sequence, else 0
# We use a reverse cumulative max, then reverse back, and shift to ensure it's a *future* event.
df['target'] = df.groupby('enodeb_id')['is_cell_unavailable_for_target'].transform(
    lambda x: x.iloc[::-1].cummax().iloc[::-1].shift(1).fillna(0)
).astype(int)

# Drop the intermediate column
df.drop(columns=['is_cell_unavailable_for_target'], inplace=True)

print("Generated 'target' variable.")
display(df.head())
```
