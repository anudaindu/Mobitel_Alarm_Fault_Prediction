Converting date time columns to datetimeobjects.

```python
df['occurred_on'] = pd.to_datetime(df['occurred_on'], errors='coerce')
df['cleared_on'] = pd.to_datetime(df['cleared_on'], errors='coerce')

# Display the DataFrame info to verify the data types
display(df.info())
```

```python
df['received_on'] = pd.to_datetime(df['received_on'], errors='coerce')
df['acknowledged_on'] = pd.to_datetime(df['acknowledged_on'], errors='coerce')

# Display the DataFrame info again to verify all datetime conversions
display(df.info())
```

Standardizing the column names

```python
import re

def standardize_column_name(col_name):
    # Convert to lowercase
    col_name = col_name.lower()
    # Replace spaces with underscores
    col_name = col_name.replace(' ', '_')
    # Remove any characters that are not alphanumeric or underscores
    col_name = re.sub(r'[^a-z0-9_]', '', col_name)
    return col_name

df.columns = [standardize_column_name(col) for col in df.columns]

print("Standardized column names:")
display(df.columns.tolist())
```

Data Cleaning

Duplication checking and removing - no duplications are found

```python
initial_rows = df.shape[0]
df.drop_duplicates(inplace=True)
duplicates_removed = initial_rows - df.shape[0]

print(f"Number of duplicate rows removed: {duplicates_removed}")
print(f"New DataFrame shape: {df.shape}")
display(df.head())
```

Handling missing values

### Evaluating Missing Values with '-' as Missing Data

As suggested, we will now consider '-' as a missing value and replace it with `NaN` across the DataFrame before re-checking for missing values.

```python
# Replace all '-' values with NaN in the DataFrame
df.replace('-', pd.NA, inplace=True)

# Re-calculate missing values after replacing '-' with NaN
missing_values_rechecked = df.isnull().sum()
missing_values_rechecked = missing_values_rechecked[missing_values_rechecked > 0].sort_values(ascending=False)

print("Columns with missing values (including '-' as NaN) and their counts:")
display(missing_values_rechecked)

# Display the DataFrame info again to see updated counts
display(df.info())
```

### Droping irrelevant columns based on Threshold

we will now drop columns that have more than 95% missing values, with the exception of `gnodeb_id` which is to be retained. For the remaining columns with missing values, including `gnodeb_id`, we will fill them appropriately.


`gnodeb_id` is required for 5G connection detections therefore it is retained.

```python
total_rows = df.shape[0]
missing_percentage = (df.isnull().sum() / total_rows) * 100

# Identify columns to drop (more than 95% missing) but exclude 'gnodeb_id'
columns_to_drop = missing_percentage[missing_percentage > 95].index.tolist()

if 'gnodeb_id' in columns_to_drop:
    columns_to_drop.remove('gnodeb_id')
    print("Retaining 'gnodeb_id' as per user request, even though it has >95% missing values.")

if columns_to_drop:
    df.drop(columns=columns_to_drop, inplace=True)
    print(f"Dropped columns with >95% missing values (excluding gnodeb_id): {columns_to_drop}")
else:
    print("No columns to drop based on the >95% missing value threshold (excluding gnodeb_id).")

# Now, re-evaluate missing values and fill them for remaining columns
# For object columns, fill with 'Unknown'
for col in df.select_dtypes(include=['object']).columns:
    if df[col].isnull().any():
        df[col].fillna('Unknown', inplace=True)

# For other types (e.g., numeric), we might need different strategies, but for now assuming object types cover the rest.
# If there are datetime columns with NaT, they will remain as such, which is acceptable.

print("\nMissing values after dropping columns and filling object types:")
missing_values_final = df.isnull().sum()
missing_values_final = missing_values_final[missing_values_final > 0].sort_values(ascending=False)

if missing_values_final.empty:
    print("All missing values have been handled.")
else:
    display(missing_values_final)

display(df.head())
```
