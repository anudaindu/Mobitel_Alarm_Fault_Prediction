### Combining Multiple Excel Files

combining excel files in mobi_data2 file to get a vast amount of combined data to analysis

```python
import os

# Define the directory where your Excel files are located
data_directory = '/content/drive/MyDrive/MobitelData/mobi_data2'

# List all files in the directory that end with .xlsx
excel_files = [f for f in os.listdir(data_directory) if f.endswith('.xlsx')]
excel_files.sort() # Optional: Sort files if order matters (e.g., by date in filename)

# Initialize an empty list to store individual DataFrames
all_dfs = []

print(f"Found {len(excel_files)} Excel files in '{data_directory}':")

# Loop through each Excel file and append its data to the list
for i, file_name in enumerate(excel_files):
    file_path = os.path.join(data_directory, file_name)
    try:
        # Read the Excel file, assuming the actual header starts at row index 4 (5th row)
        df_temp = pd.read_excel(file_path, header=4)
        all_dfs.append(df_temp)
        print(f"  - Successfully loaded '{file_name}' (shape: {df_temp.shape})")
    except Exception as e:
        print(f"  - Error loading '{file_name}': {e}")

# Concatenate all DataFrames in the list into a single DataFrame
if all_dfs:
    combined_df = pd.concat(all_dfs, ignore_index=True)
    print(f"\nSuccessfully combined {len(all_dfs)} files into a single DataFrame.")
    print("Combined DataFrame shape:", combined_df.shape)
    print("First 5 rows of the combined DataFrame:")
    display(combined_df.head())
else:
    print("No Excel files were successfully loaded and combined.")

# Update the main 'df' variable to hold the combined DataFrame
df = combined_df
```
